```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold
from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_csv("k_fold_linear_regression.csv", sep='\t')
print(df)
```

```python
df.columns = df.columns.str.replace('\t', '').str.strip()
x = df[["Hours_studied"]]
y = df[["Marks"]]

print("x: ")
print(x)
print("y: ")
print(y)
```

```python
kf=KFold(n_splits=5,shuffle=False)
print("Number of folds: ",kf.n_splits)
```

```python
mse_values = []
fold = 1
for train_index, validation_index in kf.split(x):
  x_train = x.iloc[train_index]
  y_train = y.iloc[train_index]
  x_validation = x.iloc[validation_index]
  y_validation = y.iloc[validation_index]

  model = LinearRegression()
  model.fit(x_train, y_train)
  slope = model.coef_[0]
  intercept = model.intercept_
  y_pred = model.predict(x_validation)
  mse = mean_squared_error(y_validation, y_pred)
  mse_values.append(mse)
  print("========== Fold", fold, "==========")
  print("Training records:", list(train_index))
  print("Validation records:", list(validation_index))
  print("Slope: ", round(slope[0], 4))
  print("Intercept: ", round(intercept[0], 4))
  print("Regression Equation: ")
  print("Y =", round(intercept[0], 4), "+", round(slope[0], 4), "* X")
  print("Actual Values: ", list(y_validation))
  print("Predicted Values: ", [round(value[0], 4)for  value in y_pred])
  print("Validation MSE :",round(mse,4))
  print()

  fold += 1
```

```python
from numpy import average
print("MSE for each fold: ")
for i, mse in enumerate(mse_values, start=1):
  print("Fold", i, "MSE -", round(mse, 4))
  average_mse = average(mse_values)/len(mse_values)
print("Average Cross-Validation MSE: ", round(average_mse, 4))
```
