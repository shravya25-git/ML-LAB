```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
df=pd.read_csv("linear_reggression_lab_14_record.csv", delim_whitespace=True)
df.columns = ['Student', 'Hours_Studied', 'Marks']
x=df[["Hours_Studied"]]
y=df["Marks"]
```

```python
x_train = x.iloc[:12]
x_test = x.iloc[12:]

y_train = y.iloc[:12]
y_test = y.iloc[12:]

mode1 = LinearRegression()
mode1.fit(x_train, y_train)

print("Slope:", mode1.coef_[0])
print("Intercept:", mode1.intercept_)
```

```python
train_mse = mean_squared_error(y_train, mode1.predict(x_train))
test_mse = mean_squared_error(y_test, mode1.predict(x_test))

print("Train MSE:", train_mse)
print("Test MSE:", test_mse)
```

```python
plt.scatter(x_train, y_train, color='blue', label='Train Data')
plt.scatter(x_test, y_test, color='red', label='Test Data')
plt.plot(x, mode1.predict(x), label='Regression Line', color='green')
plt.xlabel('Hours Studied')
plt.ylabel('Marks')
plt.legend()
plt.show()
```
